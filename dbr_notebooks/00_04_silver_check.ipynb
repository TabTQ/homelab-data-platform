```python
%sql
-- must return 0 rows (the dedupe worked)
SELECT metric_name, series_key, ts, count(*) c
FROM homelab.silver.prom_samples GROUP BY 1,2,3 HAVING c > 1 LIMIT 5;
```

```python
%sql
-- shape check
SELECT metric_name, count(DISTINCT series_key) series, count(*) samples,
       min(ts) first_ts, max(ts) last_ts
FROM homelab.silver.prom_samples GROUP BY 1 ORDER BY 1;
```

```python
%sql
SELECT labels['name'] AS container, count(DISTINCT series_key) AS series, min(ts), max(ts)
FROM homelab.silver.prom_samples
WHERE metric_name = 'container_mem_bytes' GROUP BY 1 ORDER BY 2 DESC;
```
